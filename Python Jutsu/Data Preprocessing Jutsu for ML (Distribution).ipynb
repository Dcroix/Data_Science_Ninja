# Business Understanding

## Business Context
A company wants to identify employees who have a high potential to become star employees. A star employee is someone who consistently demonstrates strong performance, productivity, reliability, engagement, and potential for future growth.

The company currently relies heavily on annual performance reviews and manager recommendations to identify high-potential employees. While useful, these approaches can be subjective and may overlook employees whose potential is not immediately visible.

The company would like to use historical employee data to develop a machine-learning model that estimates the probability that an employee will become a star employee.

## Business Problem
The primary business objective of this project is to help the organization identify employees who have the potential to become star employees by using existing and relevant features such as: employee performance, productivity, engagement, training, leadership, and work-related data. 

The organization aims to improve the consistency and objectivity of its talent identification process, support employee development and succession planning, prioritize training and mentoring opportunities, and improve the retention of high-potential employees. 

The resulting predictive model will serve as a decision-support tool that can provide HR managers with additional evidence when identifying employees who may benefit from career development opportunities, while still allowing human judgment to remain an important part of the decision-making process.

## Data Mining Objective
The data mining objective is to develop and evaluate a machine-learning classification model that can predict whether an employee is likely to be classified as a potential star employee based on historical employee characteristics and performance-related indicators. 

The project will involve examining the relationships between employee attributes and star-employee status, identifying the features that contribute most strongly to the prediction, and training models such as Logistic Regression, Decision Tree, Random Forest, and Extreme Gradient Boost to classify employees into star and non-star categories. The model will be evaluated using appropriate performance measures such as accuracy, precision, recall, F1-score, and ROC-AUC, with particular attention given to recall because failing to identify a genuinely high-potential employee may result in missed development and retention opportunities.

## Business Assumptions
The following are conditions that hold true based on discussions with data owners and concerned stakeholders.
- Historical employee data is reasonably representative of future employees.
- Performance-related measurements are collected consistently.
- The definition of a "star employee" is consistent across departments.
- Employee performance is influenced by measurable factors in the dataset.
- Historical labels identifying star employees are sufficiently reliable.
- The dataset does not contain substantial systematic measurement errors.
- Employee data is handled according to applicable privacy and employment regulations.
- Protected characteristics and inappropriate personal attributes should not be used as predictive features.
- The model will support, rather than replace, human decision-making.
- Model performance may change over time and therefore requires periodic monitoring.

## Success Criteria
The project will be considered successful if it helps HR:

- More consistently identify high-potential employees supported by a machine learning model with a recall of 0.85 or better.
- Prioritize development opportunities.
- Reduce the number of high-potential employees overlooked through purely subjective processes.
- Provide useful information for succession planning.

## Data Partitioning / Train - Test Split

```python
# Separate explanatory variables (X) and target variable (y)

X = df.drop(columns=["Employee_ID", "Star_Employee"])
y = df["Star_Employee"]

print("X dimensions:", X.shape)
print("y dimensions:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

# Data Pre-processing Techniques

## Standardization

```python
# Stardizing numeric explanatory variables for the train set
from sklearn.preprocessing import StandardScaler

# Only include the numeric variables
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns

# Standardize numeric variables
scaler = StandardScaler()

X_train[numeric_features] = scaler.fit_transform(
    X_train[numeric_features]
)

# View the standardized explanatory variables
display(X_train.head())
```

```python
# Now perform standardization on the test set

# Only include the numeric variables
numeric_features = X_test.select_dtypes(
    include=["int64", "float64"]
).columns

X_test[numeric_features] = scaler.transform(
    X_test[numeric_features]
)

# View the standardized explanatory variables
display(X_test.head())
```

## One-hot Coding

```python
from sklearn.preprocessing import OneHotEncoder
# Specify the categorical Variables
categorical_features = [
    "Department",
    "Job_Level"
]

# Apply one-hot coding for the TRAIN SET
encoder = OneHotEncoder(
    drop="first",
    handle_unknown="ignore",
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(
    X_train[categorical_features]
)

# Converts the results from array to a nicely labeled dataframe
X_train_encoded = pd.DataFrame(
    X_train_encoded,
    columns=encoder.get_feature_names_out(categorical_features),
    index=X_train.index
)

display(X_train_encoded.head())
```

```python
# Include the one-hot coded variables to the dataframe
X_train_final = pd.concat(
    [
        X_train.drop(columns=categorical_features),
        X_train_encoded
    ],
    axis=1
)
display(X_train_final.head())
```

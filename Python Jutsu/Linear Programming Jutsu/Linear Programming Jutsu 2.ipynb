In this notebook we will use the Pulp library from Python to help us address optimization problems. This is carried out via a process we call as linear programming. Check out the scenario below so that we can further enhance our earlier discussions.

# <font face = 'Impact' color = '#5992C6' > Optimization in Bakery Products <font/>
You are a business analysts helping out a local bakery that is known to produce:
- One batch (35 pcs) of Cupcakes (with P40 Profit) and
- One batch (35 pcs) of Ensaymada bread (with P30 Profit).  

Each of these product uses limited daily resources of the bakery. These resources are as follows:

| Item      | Cupcakes  | Ensaymada |
| --------- | --------- | --------- |
| Profit    | ₱40       | ₱30       |
| Oven Time | 2 hours   | 1 hour    |
| Flour     | 1 kg      | 1 kg      |

In a day, the bakery only has 100 hours of oven time (10 ovens working for 10 hours each in a daily basis).  
Moreover, the bakery's available flour for daily use is 80kg.  

## <font face = 'Impact' color = '#5992C6' > The Manager of the Bakery now asks you as an analysts: <font/>
### "How many Cupcakes and Ensaymada bread should the bakery produce per day to maximize profit without exceeding resources?

```python
# Load the needed library and functions
from pulp import LpMaximize, LpProblem, LpVariable
```

```python
# Create the LP model
model = LpProblem("Bakery_Production", LpMaximize)
```

```python
# Identify the Decision variables
cupcakes = LpVariable("Cakes", lowBound=0)
ensaymada = LpVariable("Bread", lowBound=0)
```

```python
# Define the model objective: maximize profit
model += 40 * cupcakes + 30 * ensaymada, "Total Profit" #40 and 30 are their corresponding profits 
```

```python
# Highlight the Constraints
model += 2 * cupcakes + 1 * ensaymada <= 100   # Oven time (hours)
model += 1 * cupcakes + 1 * ensaymada <= 80    # Flour (kg)
```

```python
# Solve using the pulp library
model.solve()
```

```python
# Sharing our Results
print("Cupcakes to produce:", cupcakes.value())
print("Ensaymada to produce:", ensaymada.value())
print("Maximum Profit:", model.objective.value())
```

## To achieve maximum profit, we recommend producing 20 batches of Cupcakes and 60 batches of ensaymada to achieve P2600 per day.

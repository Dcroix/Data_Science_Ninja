In this notebook we will use the Pulp library from Python to help us address optimization problems. This is carried out via a process we call as linear programming. Check out the scenario below so that we can further enhance our earlier discussions.

# <font face = 'Impact' color = '#5992C6' > Warehouse Distribution Cost Minimization <font/>
You are a business analyst for a company that operates 2 warehouses to supply products to 3 retail stores.  
Each warehouse has limited supply, and each store has a required demand. Shipping costs vary depending on the route.

### Supply Capacity of each Warehouse:
- Epic Zone Warehouse: 100 units
- Edge Zone Warehouse: 120 units

### Demand requirements of stores:
- Store 1: 80 units
- Store 2: 70 units
- Store 3: 60 units

### Shipping Costs
| From / To   | Store 1 | Store 2 | Store 3 |
| ----------- | ------- | ------- | ------- |
| Warehouse A | 4       | 6       | 9       |
| Warehouse B | 5       | 4       | 7       |

## <font face = 'Impact' color = '#5992C6' > The Manager of the Warehouse now asks you: <font/>
### "How should we allocate shipments from each warehouse to each store so that all customer demand is met at the lowest possible total shipping cost?"

```python
# Load the needed library and functions
from pulp import LpMaximize, LpProblem, LpVariable
```

```python
# Create the LP model
model = pulp.LpProblem("Minimize_Shipping_Cost", pulp.LpMinimize)
```

```python
# Identify the Decision variables
# x_A1 = units from Warehouse A to Store 1, etc.
x_A1 = pulp.LpVariable('x_A1', lowBound=0)
x_A2 = pulp.LpVariable('x_A2', lowBound=0)
x_A3 = pulp.LpVariable('x_A3', lowBound=0)
# x_B1 = units from Warehouse B to Store 1, etc.
x_B1 = pulp.LpVariable('x_B1', lowBound=0)
x_B2 = pulp.LpVariable('x_B2', lowBound=0)
x_B3 = pulp.LpVariable('x_B3', lowBound=0)
```

```python
# Define the model objective: minimize cost
model += (4 * x_A1 + 6 * x_A2 + 9 * x_A3 + 
          5 * x_B1 + 4 * x_B2 + 7 * x_B3)
```

```python
# Highlight the Supply Constraints
model += x_A1 + x_A2 + x_A3 <= 100  # Warehouse A
model += x_B1 + x_B2 + x_B3 <= 120  # Warehouse B

# Highlight the Demand constraints
model += x_A1 + x_B1 == 80  # Store 1
model += x_A2 + x_B2 == 70  # Store 2
model += x_A3 + x_B3 == 60  # Store 3
```

```python
# Solve using the pulp library
model.solve()
```

```python
# Sharing our Results
print("Status:", pulp.LpStatus[model.status])
for var in model.variables():
    print(f"{var.name} = {var.varValue}")
print("Total Minimum Cost =", pulp.value(model.objective))
```

## To achieve minimum costs, we recommend the following allocation of units:
- 80 units from Warehouse A shipped to Store 1
- No unit to be shipped to Store 2 from Warehouse A
- 10 units from Warehouse A shipped to Store 3
- No unit to be shipped to Store 1 from Warehouse B
- 70 units from Warehouse B shipped to Store 2
- 50 units from Warehouse B shipped to Store 3
